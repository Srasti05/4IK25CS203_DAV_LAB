Lab Experiment 4 - Data Preprocessing : Handling missing data using imputation (mean, median ,forward fill ), and encoding categorical variable

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

Step 1: Load the Dataset

```python
df = pd.read_csv("/content/student_performance_dataset.csv")
print("First 5 rows of the dataset:")
df.head()
```

```python
print(df.info())
```

Step 2: Check missing values

```python
print("\n Missing values before handling:")
print(df.isnull().sum())
```

```python
df.loc[0:5, 'Study_Hours_per_Week']=None
df.loc[10:15, 'Attendance_Rate']=None
df.loc[20:25, 'Past_Exam_Scores']=None
df.loc[30:35, 'Final_Exam_Scores']=None

print("\n Missing values after inserting sample NaNs:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

Step 3: Handle missing data using different imputation methods

Mean Imputation

```python
df['Study_Hours_per_Week']= df['Student_ID'].fillna(df['Study_Hours_per_Week'].mean())
```

Meadian Imputation

```python
df['Attendance_Rate']= df['Attendance_Rate'].fillna(df['Attendance_Rate'].median())
```

Forward Fill

```python
df['Past_Exam_Scores']=df['Past_Exam_Scores'].ffill()
```

Missing values after imputation

```python
df['Final_Exam_Scores']=df['Final_Exam_Scores'].fillna(df['Final_Exam_Scores'].mean)
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

Step 4: Encoding categorical variable

One-Hot Encoding

```python
df = pd.get_dummies(df, columns=['Parental_Education_Level'])
print("\nOne-Hot Encoded columns:")
print(df.columns.tolist())
```

Label Encoding

```python
le= LabelEncoder()
df['Pass_Fail_Encoded'] = le.fit_transform(df['Pass_Fail'])
print("\nLabel Encoded Result:")
print(df[['Pass_Fail' , 'Pass_Fail_Encoded']].head())
```

Label Encoding for other binary columns too

```python
df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df['Internet_Access_Encoded'] = le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Activities_Encoded'] = le.fit_transform(df['Extracurricular_Activities'])
```

Step 5 : Save the final cleaned dataset

```python
df.to_csv("student_performance_cleaned.csv",index=False)
print("\nCleaned dataset saved as 'student_performance_cleaned.csv'")
```
